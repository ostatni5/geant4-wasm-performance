```python

from data_parser import extract_data, extract_time
import pandas as pd

import os

import matplotlib.pyplot as plt



names_browser = ["chromium", "firefox"]
names_native = ["native", "native-multithread"]

folder = "./results/"

sub_folders = [name for name in os.listdir(folder) if os.path.isdir(os.path.join(folder, name))]

# sort the sub_folders
sub_folders.sort(key=lambda x: int(x.split('_')[1]))
print(sub_folders)


colors = plt.rcParams["axes.prop_cycle"]()

names = names_browser + names_native

def parse_output(folder):
    data_dict= {
        "names": names,
    }

    def plot(names):
        print(f'Plotting {names}')
        max_usages = []
        fig, axs = plt.subplots(len(names), 1,figsize=(5, 8), sharex=True)
        # plot each serie on subplot
        for i,name in enumerate(names):
            print(f'Plotting {name}')
            # extract data from file
            metadata, time_ram_pairs, max_usage = extract_data(folder + name + ".usage.log")
            print(f"Max ram usage for {name}: {max_usage} MiB")
            metadata["name"] = name
            max_usages.append(max_usage)
            c = next(colors)["color"]

            axs[i].plot(
                [pair[0] for pair in time_ram_pairs],
                [pair[1] for pair in time_ram_pairs],
                label=metadata["name"],
                color=c
            )
            axs[i].set_ylabel("ram [MiB]")
            axs[i].set_xlabel("time [s]")
            axs[i].set_title(metadata["name"])
            axs[i].grid(True)


        fig.tight_layout()
        fig.show()
        
        return max_usages


    def plot_ram(names):
        print(f'Plotting ram usage {names}')
        max_usages = plot(names)
        
        plt.figure(figsize=(10, 5))
        plt.bar_label(plt.bar(names, max_usages), labels=[f"{usage:.0f} MiB" for usage in max_usages])
        plt.ylabel("ram [MiB]")
        plt.xlabel("name")
        plt.title("Max ram usage [MiB]")
        plt.show()
        
        return max_usages

    max_ram_usage = plot_ram(names)
    data_dict["max_ram_usage"] = max_ram_usage


    (native_time,_) = extract_time(folder + "native.time.log")
    # ploat all in one plot
    def plot_times(names):
        print(f'Plotting times {names}')
        plt.figure(figsize=(10, 5))
        times = []
        for name in names:
            (time,_) = extract_time(folder + name + ".time.log") 
            
            times.append(time)
            
        plt.bar_label(plt.bar(names, times), labels=[f"{time:.0f}\n({(time/  native_time * 100):.0f}%)" for time in times], padding=-30)
        plt.ylabel("time [s]")
        plt.xlabel("name")
        plt.title("Time of execution")
        plt.show()
        return times
        
    time_duration = plot_times(names)

    data_dict["time_duration"] = time_duration

    df = pd.DataFrame(data_dict)
    
    df.to_csv(folder + "data.csv")
    print("Data saved to data.csv")
    
    return df


frames =[]

for sub_folder in sub_folders:
    print(f'Parsing {sub_folder}')
    df = parse_output(folder + sub_folder + "/")
    print("Parsed")
    df["bins"] = '_'.join(sub_folder.split('_')[:1])
    
    frames.append(df)
    
    
df_concat = pd.concat(frames, axis=0)

plt.figure(figsize=(10, 6))
for name in df_concat['names'].unique():
    df_plot = df_concat[df_concat['names'] == name]
    plt.plot(df_plot['bins'], df_plot['time_duration'], label=name)

plt.xlabel('bins')
plt.ylabel('time_duration')
plt.title('Time of execution')
plt.legend()
plt.show()

plt.figure(figsize=(10, 6))
for name in df_concat['names'].unique():
    df_plot = df_concat[df_concat['names'] == name]
    plt.plot(df_plot['bins'], df_plot['max_ram_usage'], label=name)

plt.xlabel('bins')
plt.ylabel('max_ram_usage')
plt.title('Max ram usage [MiB]')
plt.legend()
plt.show()
```
